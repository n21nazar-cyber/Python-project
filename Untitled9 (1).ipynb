```python
import random
import time

class RobotController:
    def __init__(self):
        self.is_moving = True

    def check_distance(self):
        # شبیه‌سازی خواندن سنسور (یک عدد تصادفی بین ۵ تا ۵۰ سانتی‌متر)
        return random.randint(5, 50)

    def move_forward(self):
        print("ربات در حال حرکت به جلو است...")

    def stop_and_turn(self):
        print("!!! مانع تشخیص داده شد! در حال توقف و چرخش...")
        self.is_moving = False  # برای مثال در این نسخه متوقف می‌کنیم

# --- منطق اصلی برنامه ---
robot = RobotController()

print("شروع فعالیت ربات...")
for i in range(5): # برای تست، ۵ بار حلقه را اجرا می‌کنیم
    dist = robot.check_distance()
    print(f"فاصله تا مانع: {dist} سانتی‌متر")
    
    if dist < 20:
        robot.stop_and_turn()
        break # پایان شبیه‌سازی پس از برخورد
    else:
        robot.move_forward()
```
